```python
import numpy as np
```

```python
seats = np.zeros((5,8),dtype = np.int64)
```

```python
print(seats)
```

```python
def display_seat():
  for i,row in enumerate(seats):
    print(f"row {i+1} : {row}")
```

```python
display_seat()
```

```python
def seat_booking(row,col):
  if row <1 or row>5 or col <1 or col >8:
    print("invalid seat!")
    return
  if seats[row-1 , col - 1] == 1:
    print("seat is already booked.")
    return
  else:
    seats[row-1,col-1]=1
    print("seat is booked.")
```

```python
seat_booking(3,4)
seat_booking(3,1)
seat_booking(3,3)
seat_booking(3,6)
seat_booking(3,7)
seat_booking(1,3)

seat_booking(2,1)
seat_booking(4,5)
seat_booking(5,4)





```
